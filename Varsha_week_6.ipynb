```python
#Gender distribution
sns.countplot(data=df, x='Gender')
plt.title("patient distribution by gender")
plt.xlabel("Gender")
plt.ylabel("Number of patience")
plt.show()
```

```python
age_bins = [-1, 17, 39, 59, float("inf")]

age_labels = [
    "Child",
    "Young Adult",
    "Middle Adult",
    "Senior"
]

df["Age Group"] = pd.cut(
    df["Age"],
    bins=age_bins,
    labels=age_labels
)
```

```python
df["Age Group"].value_counts()
```

```python
sns.countplot(data=df,x="Age Group")
plt.title("patient distribution by gender")
plt.xlabel("Age Group")
plt.ylabel("Number of patience")
plt.show()
```

```python
sns.scatterplot(
    data=df,
    x="Age",
    y="Billing Amount"
)

plt.title("Age vs Billing Amount")
plt.xlabel("Age")
plt.ylabel("Billing Amount")
plt.show()
```

```python
df["Date of Admission"].head()
```

```python
df["Date of Admission"].dtype
```

```python


df["Admission Year"] = df["Date of Admission"].dt.year
```

```python
df["Admission Month"] = df["Date of Admission"].dt.month
```

```python
df[
    [
        "Date of Admission",
        "Admission Year",
        "Admission Month"
    ]
].head()
```

```python
#Count Monthly Admissions

monthly_admissions = (
    df.groupby("Admission Month")
      .size()
      .reset_index(name="Admissions")
)
```

```python
monthly_admissions
```

```python
#Admission trend

sns.lineplot(data=monthly_admissions, x="Admission Month", y="Admissions", marker="o")
plt.title("Monthly Admissions Trend")
plt.xlabel("Month")
plt.ylabel("Number of Admissions")
plt.xticks(rotation=45)
plt.show()
```

```python
#compare admission types over time

monthly_admission_type = (
    df.groupby(
        ["Admission Month", "Admission Type"]
    )
    .size()
    .reset_index(name="Admissions")
    )
```

```python
monthly_admission_type.head()
```

```python
sns.lineplot(data=monthly_admission_type, x="Admission Month", y="Admissions", hue="Admission Type", marker="o")
plt.title("Monthly Admissions by Type")
plt.xlabel("Month")
plt.ylabel("Number of Admissions")
plt.xticks(rotation=45)
plt.show()
```

```python
df["Length of Stay"] = (df["Discharge Date"] - df["Date of Admission"]).dt.days
```

```python
sns.scatterplot(data=df, x="Length of Stay", y="Billing Amount", marker="o")
plt.title("Billing Amount vs Length of Stay")
plt.xlabel("Length of Stay")
plt.ylabel("Billing Amount")
plt.show()
```

```python
numeric_cols = [
    "Age",
    "Length of Stay",
    "Billing Amount"
]

corr = df[numeric_cols].corr()
corr
```

```python
plt.figure(figsize=(7,5))

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    vmin=-1,
    vmax=1
)

plt.title("Healthcare variables correlation matrix")
plt.show()
```
